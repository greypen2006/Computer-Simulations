```python
import numpy as np
import matplotlib.pyplot as plt
import random
```

```python
add_list=[]
```

```python
steps=100
initial_pos = 0

pos=initial_pos
for i in range(steps):
    step=np.random.choice([-1,1])
    pos=pos+step
    add_list.append(pos)

time_step = steps
```

```python
#plot
plt.plot(add_list)
plt.title("Random Walk")
plt.xlabel("Time Steps")
plt.ylabel("Position")
plt.show()
```
